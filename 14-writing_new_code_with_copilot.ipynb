# Writing New Code with Copilot

We can use Copilot to write new pieces of code. When doing so, the following are useful tips:

1. **Provide a description**: Copilot can generate code based on the description you provide. The more detailed the description, the better the code generated.
2. **Be specific**: For example, when writing a function, you might specify:
    - The types and purposes of arguments
    - The return type
    - The expected behavior
    - Any constraints or requirements
    - If and when exceptions should be raised
3. **Iterate**: Copilot may not get it right the first time. Describe problems and what you would like done differently. This can be done in chat, or by make iterative changes in the editor.
4. **Set the style**: Copilot is aware of other code in the file and can generate code that is consistent with it. Writing some code before asking Copilot to generate more can help it understand the style you are looking for.
5. **Be critical of suggestions**: Copilot may generate code that is not correct or not what you want. It is important to review the code and make sure it is correct and does what you want.
6. **Test the code**: You should always test code you write, preferably by writing formal tests in a unit-testing framework such as ```unittest``` or ```pytest``` in Python. This is doubly important when using Copilot, as the code may not be correct or may not do what you expect. Copilot can help you write tests.
7. **Understand the code**: Copilot cannot replace the need for you to understand the code you're writing. It is best suited to speeding up the production of tasks you could do yourself. It may be able to help you produce code one step more complicated than you could do yourself, but you should make sure you understand what is going on before you use it. This can be a good way to learn new techniques or patterns, and can be aided by asking Copilot for explanations of bits you don't understand. If you try to produce code that is substantially beyond your current abilities, you will likely make mistakes and not understand how to fix them. As a result, you should carry on developing your skills as a programmer instead of relying on Copilot (or any other AI tool) to do the work for you.
8. **Break down large tasks**: Copilot is better at generating small pieces of code than large ones. If you have a large task, try to break it down into smaller pieces that you can ask Copilot to generate for you. This will make it easier for Copilot to understand what you want and generate the correct code. Copilot can help you with this process by suggesting ways to break down the task.

````python
import random

def simulate_combat_game_with_abilities(player1_name, player2_name, player1_hp=100, player2_hp=100, 
                                        player1_attack=15, player2_attack=15, verbose=True):
    """
    Simulates a turn-based combat game between two players with special abilities.
    
    Each turn, players can choose to:
    - "attack": Deal normal damage (±30% variation)
    - "defend": Reduce incoming damage by 50% next turn
    - "heal": Restore 25 HP (but cannot attack that turn)
    
    The game continues until one player's HP reaches 0 or below.
    
    Args:
        player1_name (str): Name of first player
        player2_name (str): Name of second player
        player1_hp (int): Starting HP for player 1
        player2_hp (int): Starting HP for player 2
        player1_attack (int): Base attack damage for player 1
        player2_attack (int): Base attack damage for player 2
        verbose (bool): Whether to print turn details
    
    Returns:
        dict: Game result with winner, turns, final HP, and ability usage stats
    """
    # Initialize game state
    p1_hp = player1_hp
    p2_hp = player2_hp
    p1_defending = False
    p2_defending = False
    turns = 0
    max_turns = 1000
    
    # Track ability usage
    p1_abilities = {"attack": 0, "defend": 0, "heal": 0}
    p2_abilities = {"attack": 0, "defend": 0, "heal": 0}
    
    def choose_ability(current_hp, max_hp):
        """AI strategy for choosing an ability based on health."""
        if current_hp < max_hp * 0.3:  # Low health - heal or defend
            return random.choice(["heal", "defend"])
        elif current_hp < max_hp * 0.6:  # Medium health - mixed strategy
            return random.choice(["attack", "defend", "heal"])
        else:  # High health - aggressive
            return "attack"
    
    if verbose:
        print(f"\n{'='*60}")
        print(f"Combat with Abilities: {player1_name} vs {player2_name}")
        print(f"Starting HP: {player1_name}={p1_hp}, {player2_name}={p2_hp}")
        print(f"{'='*60}\n")
    
    # Game loop
    while p1_hp > 0 and p2_hp > 0 and turns < max_turns:
        turns += 1
        
        # Players choose abilities
        p1_ability = choose_ability(p1_hp, player1_hp)
        p2_ability = choose_ability(p2_hp, player2_hp)
        
        p1_abilities[p1_ability] += 1
        p2_abilities[p2_ability] += 1
        
        if verbose:
            print(f"Turn {turns}:")
            print(f"  {player1_name} chooses: {p1_ability.upper()}")
            print(f"  {player2_name} chooses: {p2_ability.upper()}")
        
        # Resolve Player 1's action
        if p1_ability == "attack":
            variance = random.uniform(0.7, 1.3)
            damage = int(player1_attack * variance)
            if p2_defending:
                damage = int(damage * 0.5)
                if verbose:
                    print(f"    → {player1_name} attacks for {damage} damage (reduced by defense)")
            else:
                if verbose:
                    print(f"    → {player1_name} attacks for {damage} damage")
            p2_hp -= damage
        elif p1_ability == "defend":
            p1_defending = True
            if verbose:
                print(f"    → {player1_name} takes a defensive stance")
        elif p1_ability == "heal":
            heal_amount = 25
            p1_hp = min(p1_hp + heal_amount, player1_hp)
            if verbose:
                print(f"    → {player1_name} heals for {heal_amount} HP")
        
        # Resolve Player 2's action
        if p2_ability == "attack":
            variance = random.uniform(0.7, 1.3)
            damage = int(player2_attack * variance)
            if p1_defending:
                damage = int(damage * 0.5)
                if verbose:
                    print(f"    → {player2_name} attacks for {damage} damage (reduced by defense)")
            else:
                if verbose:
                    print(f"    → {player2_name} attacks for {damage} damage")
            p1_hp -= damage
        elif p2_ability == "defend":
            p2_defending = True
            if verbose:
                print(f"    → {player2_name} takes a defensive stance")
        elif p2_ability == "heal":
            heal_amount = 25
            p2_hp = min(p2_hp + heal_amount, player2_hp)
            if verbose:
                print(f"    → {player2_name} heals for {heal_amount} HP")
        
        # Reset defending status for next turn
        p1_defending = False
        p2_defending = False
        
        if verbose:
            print(f"  HP: {player1_name}={max(0, p1_hp)}, {player2_name}={max(0, p2_hp)}\n")
    
    # Determine winner
    if p1_hp > 0:
        winner = player1_name
    elif p2_hp > 0:
        winner = player2_name
    else:
        winner = "Draw"
    
    if verbose:
        print(f"{'='*60}")
        print(f"Game Over! Winner: {winner} (after {turns} turns)")
        print(f"Final HP: {player1_name}={max(0, p1_hp)}, {player2_name}={max(0, p2_hp)}")
        print(f"\nAbility Usage:")
        print(f"  {player1_name}: {p1_abilities}")
        print(f"  {player2_name}: {p2_abilities}")
        print(f"{'='*60}\n")
    
    return {
        "winner": winner,
        "turns": turns,
        "player1_final_hp": max(0, p1_hp),
        "player2_final_hp": max(0, p2_hp),
        "player1_name": player1_name,
        "player2_name": player2_name,
        "player1_abilities": p1_abilities,
        "player2_abilities": p2_abilities
    }

# Example usage
result = simulate_combat_game_with_abilities("Alice", "Bob")
print(f"Result: {result}")
````
